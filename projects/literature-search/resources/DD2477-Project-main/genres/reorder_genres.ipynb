```python
import pandas as pd
import ast
```

```python
df = pd.read_csv('zeroshotclassification/zeroshot_inference.csv', index_col=0)
df['Labels'] = df['Labels'].apply(ast.literal_eval)
df['Scores'] = df['Scores'].apply(ast.literal_eval)
```

```python
def reorder_labels_scores(row):
    labels = row['Labels']
    scores = row['Scores']
    
    # Zip, sort by label, unzip
    combined = sorted(zip(labels, scores), key=lambda x: x[0])
    sorted_labels = [x[0] for x in combined]
    sorted_scores = [x[1] for x in combined]
    # sorted_labels, sorted_scores = zip(*combined)
    
    return pd.Series({'Labels': list(sorted_labels), 'Scores': list(sorted_scores)})

df[['Labels', 'Scores']] = df.apply(reorder_labels_scores, axis=1)
```
